# Welcome to the AI Agentics Course

Welcome! In this course, we will discover, step by step, the basic building blocks of **agentic AI**: how to talk to an LLM, how to give it tools, how to chain several LLMs together, and eventually how to let an agent act on its own to complete a task. Each part of the course builds on the previous one, and we will use a running example — a philosophy professor who writes an assignment, has it answered by several AI "students", and then grades the results — to make the concepts concrete.

## What is an LLM?

A **Large Language Model (LLM)** is a program trained on huge amounts of text that has learned to predict, one piece of word (a *token*) at a time, what is likely to come next. Given a prompt, it generates a response by repeatedly predicting the next token until it produces a full answer. It has no memory between separate requests, no access to the outside world, and no built-in ability to act — it only reads text and produces text. Everything we call "agentic AI" is built on top of this simple mechanism: we structure prompts, chain calls, and connect LLMs to tools and to each other so that this text-in/text-out engine can carry out more useful, multi-step tasks.

Well-known examples include GPT (OpenAI), Gemini (Google), Claude (Anthropic), and open-source models such as Mistral or Llama that you can run yourself with Ollama. In this course we will access several of them, always through the same kind of API.

## Course roadmap

* **Part 1 — Setting up the tools**: installing `uv` and Ollama, getting API keys for the University of Rennes and Google Gemini, and understanding how to query different LLM servers through the same OpenAI-compatible API.
* **Part 2 — The network professor**: our first real *agents*, built with the OpenAI Agents SDK — running several LLMs concurrently with `asyncio`, giving an agent a tool to act on the outside world (sending a push notification), wiring up a chat interface with Gradio, and chaining agents together into an actual agentic workflow.
* **Part 3 — Tokens and Byte-Pair Encoding**: what a token actually is, why LLMs work on neither whole words nor individual characters, and how the Byte-Pair Encoding (BPE) algorithm builds a model's vocabulary of tokens — writing a tiny BPE trainer ourselves, and looking at how a real, production tokenizer splits actual text.
* **Part 4 — Embeddings**: how a model turns each token id into a dense vector that captures meaning, measuring similarity between tokens with cosine similarity and Euclidean distance, uncovering relationships between words as vector arithmetic (word analogies), and opening up the neural network itself — its architecture, how training assigns its weights, and what a model's name (135M, 7B...) and quantization actually mean for its size.
* **Part 5 — The Nobel Prize**: what a frozen knowledge cutoff actually means in practice, hand-coding a live Wikipedia fetch to get around it, and wrapping that fetch as a `@function_tool` so an actual agent can decide for itself when to use it — contrasting an algorithmic workflow with a real agent, and being honest about how unreliable that decision still is on a small local model.
* **Part 6 — The TAF advisor**: MCP (Model Context Protocol) as a standard way to package tools — filesystem, browser, and memory servers — giving an agent real-time web information and persistent memory across separate conversations, then deliberately building a biased advisor to show that clean tools and a clean pipeline don't guarantee an honest agent, and closing with the same algorithmic-vs-agentic comparison played out live, with a real Pushover notification at the end.
* **Part 7 — RAG_time**: extending token embeddings to whole-sentence ones by mean-pooling a model's hidden states, comparing a homemade version against a real embeddings API and a local model actually trained for the job — and, underneath it all, what self-attention does inside a model to make a token's vector context-dependent in the first place.
* **Part 8 — Chunk Happens**: why a single embedding vector can't represent a whole document, and how chunking it into many smaller vectors fixes that — building a local, cost-free index over the 218-page IoT book, retrieving only the passages a question needs, and giving an agent a search tool over them, then redoing the same pipeline with LangChain and a real Chroma database instead of hand-written torch.
* **Part 9 — Reality Bites**: the same RAG machinery, but on a corpus an agent collects for real from the open web instead of one clean book — where retrieval quality drops noticeably, and diagnosing why (chunking and extraction, not the model) is the real lesson.

Alongside these parts, [`Appendix_models.ipynb`](Appendix_models.ipynb) is a standalone reference: it scans every part's notebook for the models they actually call and looks each one up for its real architecture facts, so it stays accurate however the course evolves rather than needing to be kept in sync by hand.
